# Data installation & inspect

```python
from keras.datasets import mnist

(x_train, y_train), (x_test, y_test)  = mnist.load_data()
```

### Exploration

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

### plotting images

```python
import matplotlib.pyplot as plt

def show_img(index):
    plt.imshow(x_train[index] ,cmap='gray')
    plt.show()
    print(f'label : {y_train[index]}')

show_img(100)
```

```python
x_train[0].min() , x_train[0].max() # range of each pixel value
```

# Preprocessing

```python
# Normalization 1-255 -> 0-1

import numpy as np

x_train=x_train.astype("float32")/255.0
x_test=x_test.astype("float32")/255.0

print(x_train.dtype)
print(f'{x_train.min()} - {x_train.max()}')
```

```python
x_train = np.expand_dims(x_train , -1)
x_test = np.expand_dims(x_test , -1)

# from (60000,28,28) to (60000,28,28,1)
# expanding (height , width) to (height , width , channels)
```

```python
print("X_train:", x_train.shape)
print("y_train:", y_train.shape)
print("X_test:", x_test.shape)
print("y_test:", y_test.shape)

print("Pixel range:", x_train.min(), "→", x_train.max())
```

### Check Class Distribution

```python
classes , counts = np.unique(y_train , return_counts=True)

print(f'Classes {classes}')
print(f'Counts {counts}')
```

```python
# better visual for classes with respect to counts
plt.bar(classes, counts)

plt.xlabel("Digit")
plt.ylabel("Number of samples")
plt.title("MNIST Class Distribution")

plt.show()
```
